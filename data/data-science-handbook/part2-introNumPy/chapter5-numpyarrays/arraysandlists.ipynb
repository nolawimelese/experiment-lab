Reviewing python lists, contiguous arrays, and NumPy arrays

```python
import numpy as np
```

```python
# create list of zeroes
print(np.zeros(10, dtype = int))

# create 2d list of zeroes
print(np.zeros((4,4)))

# create incremeting list
print(np.arange(0,20))

# create array of values equally spaced in a range
print(np.linspace(0,5,5))
```

```python
x1 = np.ones((8,8))
print(x1)
x1_sub = x1[3:6, 4:7]
x1_sub[0,0] = 30
x1_sub[0,1] = 45
x1_sub[1,0] = 32
x1_sub[1,1] = 12
print(x1)
```

```python
# reshape array

x = np.zeros(10)
print(x)
print(x.reshape((2,5)))
```
