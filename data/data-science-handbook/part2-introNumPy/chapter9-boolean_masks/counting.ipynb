```python
import numpy as np
import seaborn as sns

rng = np.random.default_rng()
print(sns.get_dataset_names())
```

```python
df = sns.load_dataset('car_crashes').head(10)
print(df)
len(df)
```

```python
%matplotlib inline
import matplotlib.pyplot as plt

sns.set_theme(style='whitegrid')

plt.figure(figsize=(8, 4.5))
sns.barplot(data=df, x='abbrev', y='total', color='#4269d0', width=0.65)

plt.xlabel('State')
plt.ylabel('Total crashes per billion miles')
plt.title('Total car crashes')
sns.despine(left=True)

plt.show()
```

```python
x = sns.load_dataset('car_crashes')['total']

print(x < 15)
```

```python
# states with less than 15 car crashes per billion miles
print(np.count_nonzero(x < 15))
```

masking

```python
rand_ints = rng.integers(0, 10, (5,5))
print(rand_ints)
print(rand_ints < 5)
```

```python
# masking

rand_ints[rand_ints < 5]
```

remember to use & and | for boolean operations on numpy arrays. Using and/or may give unwanted results
